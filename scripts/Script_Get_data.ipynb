```python
import sys
import urllib.error
import urllib.request
from pathlib import Path

# Ensure target raw data directory exists
RAW_DIR = Path("F:/GitHub_Repos/AVIATION/data/raw_data")
#RAW_DIR.mkdir(parents=True, exist_ok=True)

# Eurostat open REST endpoints (SDMX-CSV format, no private key required)
DATASETS_TO_DOWNLOAD = {
    "eurostat_aviation_ghg_emissions.csv": (
        "https://ec.europa.eu/eurostat/api/dissemination/sdmx/2.1/data/"
        "env_ac_ainah_r2/?format=SDMX-CSV&compressed=false&nace_r2=H51&airpol=GHG"
    ),
    "eurostat_aviation_traffic_pax.csv": (
        "https://ec.europa.eu/eurostat/api/dissemination/sdmx/2.1/data/"
        "avia_paoc/?format=SDMX-CSV&compressed=false&unit=PAS"
    ),
}


def download_data():
    print("=" * 65)
    print("ASSIGNMENT A2: DOWNLOADING CANDIDATE DATA INTO data/raw/")
    print("=" * 65)

    headers = {"User-Agent": "A2DataPlanStudent/1.0 (Matriculation_6385648)"}

    for filename, url in DATASETS_TO_DOWNLOAD.items():
        destination = RAW_DIR / filename
        print(f"\n[INFO] Checking: {filename}")

        if destination.exists() and destination.stat().st_size > 1024:
            print(
                f"[EXISTS] File already present ({destination.stat().st_size} bytes). Skipping."
            )
            continue

        print(f"[DOWNLOADING] Querying endpoint: {url}")
        req = urllib.request.Request(url, headers=headers)
        try:
            with (
                urllib.request.urlopen(req, timeout=60) as response,
                open(destination, "wb") as f_out,
            ):
                f_out.write(response.read())
            print(
                f"[SUCCESS] Saved to {destination} ({destination.stat().st_size} bytes)"
            )
        except urllib.error.URLError as e:
            print(f"[ERROR] Failed to download {filename}: {e}", file=sys.stderr)

    print("\n[COMPLETE] All candidate files verified in data/raw/")


if __name__ == "__main__":
    download_data()
```
